```python
console.log("First Line");
setTimeout((a=5,b=6)=>{
    console.log("Second Line");
    console.log(a+b);
},10000);
console.log("Last Line");
```

# Promises

```python
function calculate(a,b){
    return new Promise((resolve,reject)=>{
        let sum = a+b;
        if(sum>10){
            resolve("Number is in the Range");
        }else{
            reject("Number belongs to Invalid Range");
        }
    });
}
calculate(4,4).then((m)=>{
    console.log(m);
},(e)=>{
    console.log(e);
});
```

# catch()

```python
let myPromise = new Promise(function(resolve,reject){
    let ok = false;
    if(ok){
        resolve("Fine");
    }else{
        reject("Not Fine");
    }
});
myPromise.then((value)=>{
    console.log(value);
})
.catch((value)=>{
    console.log("Error Occured");
    console.log(value);
});
```

# Reurning a Promise
Returning a promise from then()

```python
function step1(){
    return Promise.resolve("A");
}
function step2(value){
    return Promise.resolve(value+"B");
}
function step3(value){
    return Promise.resolve(value+"C");
}
function step4(value){
    return Promise.resolve(value+"D");
}
function step5(value){
    return Promise.resolve(value+"E");
}
function step6(value){
    return Promise.resolve(value+"F");
}

//Run these 6 functions step by step. 
step1()
.then((v)=>{
    console.log(v);
    return step2(v);
})
.then((v)=>{
    console.log(v);
    return step3(v);
})
.then((v)=>{
    console.log(v);
    return step4(v);
})
.then((v)=>{
    console.log(v);
    return step5(v);
})
.then((v)=>{
    console.log(v);
    return step6(v);
})
.then((v)=>{
    console.log(`Final Resultant String : ${v}`);
    console.log("Bye");
})
.catch((e)=>{
    console.log("error");
});
```

```python
let mp = new Promise(function(resolve,reject){
    let flag = true;
    setTimeout(()=>{
        if(flag){
            console.log("Success");
            resolve("Correct Value")
        }else{
            console.log("Failure");
            reject("Wrong Value");
        }
    });
});
mp
.then((value)=>{
    console.log(`Value Received : ${value}`);
})
.catch((value)=>{
    console.log(`Value Received : ${value}`);
})
.finally(()=>{
    console.log(`Task Finished. It was either true or false`);
});
```

# promise.all()

```python
function getUser(id){
    return new Promise((resolve)=>{
        resolve({name:"Husnain Ahmad"});
    });
}
function getUserPosts(id){
    return new Promise((resolve)=>{
        resolve(["Post 1","Post 2","Post 3","post 4","Post 5"]);
    });
}
Promise.all([getUser(1),getUserPosts(1)])
.then((results)=>{
    const user = results[0];
    const posts = results[1];
    console.log(user);
    console.log(posts);
})
.catch((error)=>{
    console.log("Something wrong");
});
```

In above example we used results array to get the result of all promises. We can also use destructuring which is more best and neat.

```python
function getUser(id){
    return new Promise((resolve)=>{
        resolve({name:"Husnain Ahmad"});
    });
}
function getUserPosts(id){
    return new Promise((resolve)=>{
        resolve(["Post 1","Post 2","Post 3","post 4","Post 5"]);
    });
}
Promise.all([getUser(1),getUserPosts(1)])
.then(([user, posts])=>{
    console.log(user);
    console.log(posts);
})
.catch((err)=>{
    console.log(err);
});
```

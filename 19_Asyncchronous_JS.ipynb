# setTimeout() -- an asynchronous function
It is a built-in function used to execute a piece of code or a function after a specified delay (measured in millisecods)

(1000ms = 1second)

```typescript
console.log("First Line");
setTimeout((a=5,b=6)=>{
    console.log("Second Line");
    console.log(a+b);
},10000);
console.log("Last Line");
```

# Callback

```typescript
function myDisplay(res){
    console.log(`Result : ${res}`);
}
function calculate(num1,num2,callback){
    let sum = num1+num2;
    callback(sum);
}
calculate(5,6,myDisplay);
```

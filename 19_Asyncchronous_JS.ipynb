# setTimeout() -- an asynchronous function
It is a built-in function used to execute a piece of code or a function after a specified delay (measured in millisecods)

(1000ms = 1second)

```typescript
console.log("First Line");
setTimeout((a=5,b=6)=>{
    console.log("Second Line");
    console.log(a+b);
},10000);
console.log("Last Line");
```

# Callback

```typescript
function myDisplay(res){
    console.log(`Result : ${res}`);
}
function calculate(num1,num2,callback){
    let sum = num1+num2;
    callback(sum);
}
calculate(5,6,myDisplay);
```

# Promises

```typescript
function calculate(a,b){
    return new Promise((resolve,reject)=>{
        let sum = a+b;
        if(sum>10){
            resolve("Number is in the Range");
        }else{
            reject("Number belongs to Invalid Range");
        }
    });
}
calculate(4,4).then((m)=>{
    console.log(m);
},(e)=>{
    console.log(e);
});
```

# catch()

```typescript
let myPromise = new Promise(function(resolve,reject){
    let ok = false;
    if(ok){
        resolve("Fine");
    }else{
        reject("Not Fine");
    }
});
myPromise.then((value)=>{
    console.log(value);
})
.catch((value)=>{
    console.log("Error Occured");
    console.log(value);
});
```

# Reurning a Promise
Returning a promise from then()

```typescript
function step1(){
    return Promise.resolve("A");
}
function step2(value){
    return Promise.resolve(value+"B");
}
function step3(value){
    return Promise.resolve(value+"C");
}
function step4(value){
    return Promise.resolve(value+"D");
}
function step5(value){
    return Promise.resolve(value+"E");
}
function step6(value){
    return Promise.resolve(value+"F");
}

//Run these 6 functions step by step. 
step1()
.then((v)=>{
    console.log(v);
    return step2(v);
})
.then((v)=>{
    console.log(v);
    return step3(v);
})
.then((v)=>{
    console.log(v);
    return step4(v);
})
.then((v)=>{
    console.log(v);
    return step5(v);
})
.then((v)=>{
    console.log(v);
    return step6(v);
})
.then((v)=>{
    console.log(`Final Resultant String : ${v}`);
    console.log("Bye");
});
```

# Write a function called checkAge(age) that returns a Promise. The Promise should:

Resolve with "✅ Access granted!" if age is 18 or above

Reject with "❌ Access denied! Too young." if age is below 18

Test it with checkAge(20) and checkAge(15) — handle both outcomes with .then() and .catch()

```typescript
function checkAge(age){
    return new Promise((resolve,reject)=>{
        if(age>=18){
            resolve("Access Granted");
        }else{
            reject("Access Denied");
        }
    });
}
checkAge(17)
.then((msg)=>{
    console.log(msg);
})
.catch((msg)=>{
    console.log(msg);
});
```

# Write a function loginUser(username, password) that returns a Promise:

Wait 1.5 seconds to simulate a server call

If username is "admin" AND password is "1234" → resolve with { username: "admin", role: "superuser" }

Otherwise → reject with "❌ Invalid credentials"

Test with correct and wrong credentials. Handle both with .then() and .catch()

```typescript
function loginUser(username,password){
    return new Promise((resolve,reject)=>{
        setTimeout(()=>{
            if((username === "admin") && (password === 1234)){
                resolve({username:"admin",role:"superuser"});
            }else{
                reject("Invalid Credentials");
            }
        },1500)
    })
}
loginUser("admin",123)
.then((res)=>{
    console.log(res);
})
.catch((res)=>{
    console.log(res);
});
```

Chain three .then() calls:

First .then() → multiply by 2, return the result

Second .then() → add 10, return the result

 # Write two functions:

getOrder() → returns a Promise that resolves after 1s with { orderId: 101, item: "Headphones", price: 150 }

applyDiscount(order) → returns a Promise that resolves after 500ms with the same order object but price reduced by 20%

Chain them together so that:

You get the order

You pass it to applyDiscount()

You log the final discounted order

```typescript
function getOrder(){
    return new Promise((resolve,reject)=>{
        setTimeout(()=>{
            resolve({orderId:101,item:"Headphones",price:150});
        },1000);
    });
}
function applyDiscount(order){
    return new Promise((resolve,reject)=>{
        setTimeout(()=>{
            const discountedPrice = order.price*0.8;
            resolve({...order, price:discountedPrice});
        },500);
    });
}
getOrder()
.then((order)=>{
    console.log(order);
    return applyDiscount(order);
})
.then((finalOrder)=>{
    console.log(finalOrder);
});
```

# Exercise 7 — Basic Error Handling
Write a function divide(a, b) that returns a Promise:

If b === 0 → reject with "❌ Cannot divide by zero"

Otherwise → resolve with the result

Chain .then() to log the result and .catch() to log the error.

Test with divide(10, 2) and divide(10, 0)

```typescript
function divide(a,b){
    return new Promise((resolve,reject)=>{
        if(b===0){
            reject("Cannot divide by zero");
        }else{
            resolve(a/b);
        }
    });
}
divide(10,0)
.then((result)=>{
    console.log("Result : "+result);
})
.catch((error)=>{
    console.log("Error : "+ error);
});
```

# Exercise 9 — Full Simulation
Write a function fetchData(shouldFail) that returns a Promise:

If shouldFail is true → reject after 1s with "🔴 Server error!"

If shouldFail is false → resolve after 1s with { data: "Here is your data! 📦" }

Build a complete chain with:

.then() that logs the data

Another .then() that logs "✅ Data processed successfully"

.catch() that logs the error

Test both fetchData(false) and fetchData(true) and make sure the right paths run.

```typescript
function fetchData(shouldFail){
    return new Promise((resolve,reject)=>{
        setTimeout(()=>{
            if(shouldFail){reject("Server Error");}
            else{resolve({data:"Here is Your Data"});}
        },1000);
    });
}
fetchData(false)
.then((res)=>{
    console.log("Data : ",res);
})
.then(()=>{
    console.log("Data Processed Successfully");
})
.catch((err)=>{
    console.log("Error Caught : "+err);
});
```

# Exercise 10 — Basic Promise.all()
Create three separate promises:

p1 → resolves after 1 second with "🐶 Dog"

p2 → resolves after 2 seconds with "🐱 Cat"

p3 → resolves after 500ms with "🐹 Hamster"

Use Promise.all() to wait for all three and log each animal.

Question to think about: How long will the whole thing take? Write your guess as a comment before running it!

```typescript
let p1 = new Promise((resolve,reject)=>{
    setTimeout(()=>{resolve("Dog");},1000);
});
let p2 = new Promise((resolve,reject)=>{
    setTimeout(()=>{resolve("Cat");},2000);
});
let p3 = new Promise((resolve,reject)=>{
    setTimeout(()=>{resolve("Hamster");},500);
});
Promise.all([p1,p2,p3])
.then(([first,second,third])=>{
    console.log(`First Animal ${first}`);
    console.log(`Second Animal ${second}`);
    console.log(`Third Animal ${third}`);
});
```

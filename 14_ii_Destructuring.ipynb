# Destructuring
allows to unpack values from an array into individual variablesusing their index positions, written in a concise readable way.

To skip a value, use comma with no variable name

```typescript
const arr = [1,2,3,4,5];
const [one,two,,,five] = arr;
console.log(one);
console.log(two);
console.log(five);
```
